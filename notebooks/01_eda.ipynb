# CropVision — EuroSAT RGB exploratory analysis

This notebook reads the validated, deterministic manifests produced by `src.data`. It does not download data or create a new split. Run the commands in `SETUP.md` first.

```python
from collections import Counter
from pathlib import Path

from src.preprocessing import read_manifest, plot_class_distribution, plot_example_grid
from src.utils import load_config

config = load_config('../configs/default.yaml')
split_dir = Path('..') / config['data']['split_dir']
dataset_dir = Path('..') / config['data']['dataset_dir']
records = []
for split_name in ('train', 'validation', 'test'):
    records.extend(read_manifest(split_dir / f'{split_name}.csv'))
len(records), Counter(record.label for record in records)
```

```python
figure_dir = Path('..') / config['eda']['output_dir']
plot_class_distribution(records, figure_dir / 'class_distribution.png', config['eda']['figure_dpi'])
```

```python
plot_example_grid(
    records, dataset_dir, figure_dir / 'example_grid.png',
    examples_per_class=config['eda']['examples_per_class'],
    seed=config['project']['seed'], dpi=config['eda']['figure_dpi'],
)
```
